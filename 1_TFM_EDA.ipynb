# 1. Inspección Inicial de los datos

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os

file_path = '/content/BBDD_sintetico_consistente.xlsx'

# Diagnóstico del archivo
if os.path.exists(file_path):
    file_size = os.path.getsize(file_path)
    print(f"El archivo existe. Tamaño: {file_size} bytes ({file_size / 1024:.2f} KB)")

    # Intentar leer los primeros bytes para ver si es un archivo de texto
    try:
        with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
            first_line = f.readline()[:200]
            print(f"Primeros caracteres del archivo: {repr(first_line)}")
    except Exception as e:
        print("No se pudo leer como texto plano.")
else:
    print("El archivo no existe en la ruta especificada.")

# Intento de lectura estándar (comenta esta línea para realizar solo la prueba si vuelve a fallar)
df = pd.read_excel(file_path)
display(df.head())
```

```python
print(f"El DataFrame tiene {len(df)} registros.")
```

### Visualización: Distribución de 'Tipo Cliente'

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='Tipo Cliente', order=df['Tipo Cliente'].value_counts().index, palette='viridis', hue='Tipo Cliente', legend=False)
plt.title('Distribución de Tipos de Cliente')
plt.xlabel('Conteo')
plt.ylabel('Tipo de Cliente')
plt.tight_layout()
plt.show()
```

### Visualización: Distribución de 'Modo Envio Fact' por 'Tarifa' (Top 10)

```python
# Obtener las 10 tarifas más frecuentes
top_10_tarifas = df['Tarifa'].value_counts().head(10).index

# Filtrar el DataFrame para incluir solo estas tarifas
df_top_tarifas = df[df['Tarifa'].isin(top_10_tarifas)]

plt.figure(figsize=(14, 8))
sns.countplot(data=df_top_tarifas, y='Tarifa', hue='Modo Envio Fact', order=top_10_tarifas, palette='viridis')
plt.title('Distribución de Modo de Envío de Factura para las Top 10 Tarifas (Corregido)')
plt.xlabel('Conteo')
plt.ylabel('Tarifa')
plt.legend(title='Modo Envio Fact')
plt.tight_layout()
plt.show()
```

### Valores Únicos y sus Conteos para Columnas Categóricas

```python
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    print(f"\nColumna: {col}")
    print("Conteo de valores únicos:")
    display(df[col].value_counts().head(10)) # Muestra los 10 valores más frecuentes
    print(f"Número total de valores únicos: {df[col].nunique()}")
```

### Continuación: Valores Únicos y sus Conteos para Columnas Categóricas

```python
categorical_cols = df.select_dtypes(include=['object']).columns

# Find the index of 'Tarifa' to continue from where the previous output left off
try:
    start_index = list(categorical_cols).index('Tarifa')
    remaining_categorical_cols = categorical_cols[start_index:]
except ValueError:
    # If 'Tarifa' is not found (e.g., already processed fully or not present), process all
    remaining_categorical_cols = categorical_cols

for col in remaining_categorical_cols:
    print(f"\nColumna: {col}")
    print("Conteo de valores únicos:")
    display(df[col].value_counts().head(10))
    print(f"Número total de valores únicos: {df[col].nunique()}")
```
